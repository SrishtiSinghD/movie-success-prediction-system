```python
# ======================================================
# GENAI MOVIE SUMMARY GENERATOR USING GEMINI API
# DATASET + USER INPUT VERSION
# ======================================================

# STEP 1: Install Required Libraries
!pip install -q google-generativeai pandas

# ======================================================
# STEP 2: Import Libraries
# ======================================================

import pandas as pd
import google.generativeai as genai
import getpass
# ======================================================
# STEP 3: Configure Gemini API
# ======================================================

GEMINI_API_KEY = getpass.getpass("Enter Api key")

genai.configure(api_key=GEMINI_API_KEY)

# ======================================================
# STEP 4: Load Gemini Model
# ======================================================

model = genai.GenerativeModel("gemini-2.5-flash")

# ======================================================
# STEP 5: Load Dataset
# ======================================================

# Load CSV File
df = pd.read_excel("movies_cleaned.xlsx")

# Convert movie names to lowercase for matching
df["name_lower"] = df["name"].str.lower()

# ======================================================
# STEP 6: Function to Generate Movie Summary
# ======================================================

def generate_movie_summary(movie_name):

    movie_name_lower = movie_name.lower()

    # --------------------------------------------------
    # CASE 1 → MOVIE FOUND IN DATASET
    # --------------------------------------------------

    if movie_name_lower in df["name_lower"].values:

        movie_data = df[df["name_lower"] == movie_name_lower].iloc[0]

        # Extract Features from Dataset
        genre = movie_data["genre"]
        year = movie_data["year"]
        director = movie_data["director"]
        writer = movie_data["writer"]
        star = movie_data["star"]
        rating = movie_data["rating"]
        score = movie_data["score"]
        votes = movie_data["votes"]
        country = movie_data["country"]
        runtime = movie_data["runtime"]
        company = movie_data["company"]
        budget = movie_data["budget"]
        gross = movie_data["gross"]
        released = movie_data["released"]

        # Prompt using dataset features
        prompt = f"""
        Generate a professional and engaging movie summary
        in around 80-100 words.

        Movie Title: {movie_name}

        Genre: {genre}
        Release Year: {year}
        Released On: {released}

        Director: {director}
        Writer: {writer}

        Main Star: {star}

        Rating: {rating}
        IMDb Score: {score}
        Votes: {votes}

        Country: {country}
        Runtime: {runtime} minutes

        Production Company: {company}

        Budget: ${budget}
        Gross Collection: ${gross}

        Create a cinematic, attractive, and polished summary
        based on the above movie information.
        """

    # --------------------------------------------------
    # CASE 2 → MOVIE NOT FOUND IN DATASET
    # --------------------------------------------------

    else:

        prompt = f"""
        Generate a professional and engaging movie summary
        in around 80-100 words.

        Movie Title: {movie_name}

        Create a cinematic and attractive summary.
        """

    # Generate AI Response
    response = model.generate_content(prompt)

    return response.text

# ======================================================
# STEP 7: USER INPUT
# ======================================================

movie_name = input("Enter Any Movie Name: ")

# ======================================================
# STEP 8: Generate Summary
# ======================================================

summary = generate_movie_summary(movie_name)

# ======================================================
# STEP 9: Display Output
# ======================================================

print("\n===================================")
print("MOVIE TITLE:")
print(movie_name)

print("\nAI GENERATED SUMMARY:")
print(summary)

print("===================================")
```
