```python
# ======================================================
# GENAI MOVIE SUMMARY GENERATOR USING GEMINI API
# ======================================================

# STEP 1: Install Gemini SDK

!pip install -q google-generativeai pandas

# ======================================================
# STEP 2: Import Libraries
# ======================================================

import pandas as pd
import google.generativeai as genai
import getpass
# ======================================================
# STEP 3: Load Dataset
# ======================================================

# Upload movies.csv in Colab

movies = pd.read_csv('movies.csv')

# ======================================================
# STEP 4: Configure Gemini API
# ======================================================

GEMINI_API_KEY = getpass.getpass("Enter Api key")

genai.configure(api_key=GEMINI_API_KEY)

# ======================================================
# STEP 5: Load Gemini Model
# ======================================================

model = genai.GenerativeModel("gemini-2.5-flash")

# ======================================================
# STEP 6: Function to Generate Movie Summary
# ======================================================

def generate_movie_summary(title, genre):

    prompt = f"""
    Generate a professional and engaging movie summary
    in around 80-100 words.

    Movie Title: {title}

    Genre: {genre}


    Create a cinematic and attractive summary.
    """

    response = model.generate_content(prompt)

    return response.text

# ======================================================
# STEP 7: USER INPUT FOR MOVIE NAME
# ======================================================

movie_name = input("Enter Movie Name: ")

# ======================================================
# STEP 8: Search Movie in Dataset
# ======================================================

movie_data = movies[
    movies['name'].str.lower() == movie_name.lower()
]

# ======================================================
# STEP 9: Generate Summary
# ======================================================

if len(movie_data) > 0:

    movie = movie_data.iloc[0]

    title = movie['name']
    genre = movie['genre']

    summary = generate_movie_summary(
        title,
        genre
    )

    print("\n===================================")
    print("MOVIE TITLE:")
    print(title)

    print("\nAI GENERATED SUMMARY:")
    print(summary)
    print("===================================")

else:

    print("Movie not found in dataset.")

```
